This notebook enables the remote execution of the Data Degradation and Augmentation framework directly from Google Drive. It handles environment provisioning for Python 3.10 and executes the pipeline in non-interactive mode.

### 1️⃣ Cell 1: Google Drive Mounting & Workspace Setup

```python
from google.colab import drive
import os
drive.mount('/content/drive')

# Set path to the project folder on Drive
PROJECT_PATH = '/content/drive/MyDrive/Omnia_GP/Data_Degradation'
if os.path.exists(PROJECT_PATH):
    os.chdir(PROJECT_PATH)
    print(f"[SUCCESS] Switched workspace to: {os.getcwd()}")
else:
    print(f"[ERROR] Path {PROJECT_PATH} not found. Please check your Drive folder structure.")
```

### 2️⃣ Cell 2: Colab Environment Provisioning (Python 3.10 Downgrade)
Forcing installation of Python 3.10 and aligning dependencies natively.

```python
%%bash
sudo apt-get update -y
sudo apt-get install python3.10 python3.10-distutils -y
wget https://bootstrap.pypa.io/get-pip.py
python3.10 get-pip.py
python3.10 -m pip install --upgrade pip
python3.10 -m pip install -r requirements.txt
python3.10 -m pip install tabulate matplotlib
```

### 3️⃣ Cell 3: Autonomous Pipeline Execution Flag Trigger
Executing the full 2-Stage pipeline non-interactively via subprocess mocking.

```python
import subprocess
import sys

# Pre-defined inputs for non-interactive execution:
# 1. Expand? 'y'
# 2. Factor? '2'
# 3. Method? 'warp'
# 4. Disease? 'BL'
# 5. Joint? '0'
input_str = "y\n2\nwarp\nBL\n0\n"

print("[*] Launching Autonomous Pipeline Execution...")
process = subprocess.Popen(
    ['python3.10', 'main.py'],
    stdin=subprocess.PIPE,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

stdout_data, _ = process.communicate(input=input_str)
print(stdout_data)

if process.returncode == 0:
    print("[SUCCESS] Pipeline executed and artifacts saved to Google Drive.")
else:
    print(f"[FAILURE] Pipeline exited with code {process.returncode}")
```
